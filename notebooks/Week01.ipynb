> ### "To trust, but verify, AI-generated outputs by yourself!"

| | |
|---|---|
| **AI writes code, not correct code** | Generative AI can create code, but the code may contain errors or fail in specific situations. |
| **You need to verify it** | Basic programming knowledge allows us to test, verify, debug, and improve AI-generated code. |
| **You stay in control** | Programming skills help us use AI as an assistant while maintaining human judgment, control, and problem-solving ability. |

#### <font color="#C9C7BF" size="1">Create a New Cell</font>

<img src="https://raw.githubusercontent.com/youngkorea/ai-programming-fundamentals/main/img/w01_topic_create-a-new-cell.svg" width="900" alt="Create a New Cell">

- Press <kbd>Ctrl</kbd> + <kbd>M</kbd> first so that Colab treats the next key as a command rather than text.
- Press <kbd>A</kbd> to insert a new code cell **above** the selected cell.
- Press <kbd>B</kbd> to insert a new code cell **below** the selected cell.
